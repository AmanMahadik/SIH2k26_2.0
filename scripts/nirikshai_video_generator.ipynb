```python
import os
import asyncio
from PIL import Image
import edge_tts
from moviepy.editor import ImageClip, AudioFileClip, concatenate_videoclips

# Voice options: 'en-IN-PrabhatNeural' (Male) or 'en-IN-NeerjaNeural' (Female)
VOICE = "en-IN-PrabhatNeural"

scenes = [
    {
        "id": "scene1_intro",
        "image": r"c:/SIH2k26_2.0/public/nirikshai_platform_showcase.png",
        "script": "Namaste evaluators. We are Team Code4Bharat, presenting NirikshAI 2.0 for Smart India Hackathon Problem Statement 26261 under AICTE, Ministry of Education. Every year, over 4,700 technical colleges apply for Extension of Approval. Officers spend 45 to 60 minutes per file manually verifying documents. NirikshAI transforms this with our guiding philosophy: Check early. Prove everything. Humans decide."
    },
    {
        "id": "scene2_institution",
        "image": r"C:/Users/PC/.gemini/antigravity/brain/d9aff015-221e-41d3-a913-1f1f2fdabad4/.user_uploaded/media_1793200969061_85cc9297.png",
        "script": "In the Institution Portal, colleges run pre-submission self-scrutiny before official filing. NirikshAI delivers an instant 85% readiness score and flags expired State NOCs and land built-up area mismatches, eliminating 70% of repeated deficiency rounds. Files are hashed on-device using client-side SHA-256."
    },
    {
        "id": "scene3_officer",
        "image": r"C:/Users/PC/.gemini/antigravity/brain/d9aff015-221e-41d3-a913-1f1f2fdabad4/.user_uploaded/media_1793201433787_27b4f607.png",
        "script": "Next, the Scrutiny Officer Console. Our Direct-Anchored Evidence Canvas displays pinpoint OCR bounding boxes directly on the document text with 96% confidence. If an issuer API times out, our Seven-State Verification Machine automatically routes it to Manual Review Required—ensuring zero silent passes while keeping 100% decision authority with the officer."
    },
    {
        "id": "scene4_mobile",
        "image": r"C:/Users/PC/.gemini/antigravity/brain/d9aff015-221e-41d3-a913-1f1f2fdabad4/.user_uploaded/media_1793203292488_e03fc533.png",
        "script": "To fulfill Problem Statement Objective 5, officers use our dedicated EVC Mobile Companion App for on-site physical inspections. The camera automatically embeds real-time GPS coordinate watermarks and UTC timestamps on photos of computer labs and ramps, sealing evidence with an on-device SHA-256 hash to eliminate recycled fraud photos."
    },
    {
        "id": "scene5_auditor",
        "image": r"C:/Users/PC/.gemini/antigravity/brain/d9aff015-221e-41d3-a913-1f1f2fdabad4/.user_uploaded/media_1793201257380_294b5d31.png",
        "script": "For legal non-repudiation, our Auditor Portal records every action in a SHA-256 hash-chained ledger compliant with Section 65B of the Indian Evidence Act. Auditors can mathematically verify the entire chain, while our live tamper simulation proves that unauthorized database modifications trigger immediate alerts."
    },
    {
        "id": "scene6_cloud",
        "image": r"C:/Users/PC/.gemini/antigravity/brain/d9aff015-221e-41d3-a913-1f1f2fdabad4/.user_uploaded/media_1793201213814_b320d47a.png",
        "script": "NirikshAI is live in production on Vercel Edge, backed by an Azure PostgreSQL Flexible Server in Central India with TLS 1.3 encryption. By reducing scrutiny time from 60 minutes to under 15 seconds, NirikshAI delivers a faster, transparent, and fraud-proof future for AICTE. Thank you! Jai Hind."
    }
]

os.makedirs("temp_video", exist_ok=True)
clips = []

print("Generating AI Voiceovers and Video Clips...")
for idx, sc in enumerate(scenes):
    audio_path = f"temp_video/audio_{idx}.mp3"
    
    # Generate TTS audio
    communicate = edge_tts.Communicate(sc["script"], VOICE)
    asyncio.run(communicate.save(audio_path))
    
    audio_clip = AudioFileClip(audio_path)
    duration = audio_clip.duration + 0.5
    
    # Create Image Clip resized to 1920x1080
    img = Image.open(sc["image"]).convert("RGB")
    img_resized_path = f"temp_video/img_{idx}.png"
    img.resize((1920, 1080), Image.Resampling.LANCZOS).save(img_resized_path)
    
    clip = ImageClip(img_resized_path).set_duration(duration).set_audio(audio_clip)
    clips.append(clip)
    print(f"  ✓ Scene {idx+1} completed ({duration:.1f}s)")

# Concatenate all scenes into final video
print("\nRendering final video: nirikshai_demo.mp4...")
final_video = concatenate_videoclips(clips, method="compose")
final_video.write_videofile("nirikshai_demo.mp4", fps=24, codec="libx264", audio_codec="aac")
print("\n🎉 Video successfully exported to nirikshai_demo.mp4!")
```
